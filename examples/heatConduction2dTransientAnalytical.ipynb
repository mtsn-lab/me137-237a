# 2D Transient Heat Conduction In a Homogeneous Material

Consider a 2D rectangular domain $\Omega = \{ (x,y) \in [0, L] \times [0, H] \}$ with constant thermal conductivity $k$, density $\rho$, and specific heat capacity $c_p$. The thermal diffusivity is defined as:

$$\alpha = \frac{k}{\rho c_p}$$

The 2D transient heat conduction equation without internal heat generation is governed by the parabolic partial differential equation:

$$\frac{\partial T(x,y,t)}{\partial t} = \alpha \left( \frac{\partial^2 T(x,y,t)}{\partial x^2} + \frac{\partial^2 T(x,y,t)}{\partial y^2} \right)$$

To obtain a well-posed problem, we define homogeneous Dirichlet boundary conditions (all edges held at reference temperature $T_0$) and a uniform initial temperature distribution $T(x,y,0) = T_i$:

$$T(0, y, t) = T_0, \quad T(L, y, t) = T_0 \quad (0 \le y \le H, \; t > 0)$$

$$T(x, 0, t) = T_0, \quad T(x, H, t) = T_0 \quad (0 \le x \le L, \; t > 0)$$

$$T(x, y, 0) = T_i \quad (0 < x < L, \; 0 < y < H)$$

To simplify the mathematical formulation, define the relative temperature difference $\theta(x,y,t) = T(x,y,t) - T_0$. The PDE and boundary/initial conditions transform to:

$$\frac{\partial \theta}{\partial t} = \alpha \left( \frac{\partial^2 \theta}{\partial x^2} + \frac{\partial^2 \theta}{\partial y^2} \right)$$

$$\theta(0, y, t) = 0, \quad \theta(L, y, t) = 0, \quad \theta(x, 0, t) = 0, \quad \theta(x, H, t) = 0$$

$$\theta(x, y, 0) = \theta_0 = T_i - T_0$$

Assume a product solution where spatial variables ($x, y$) and time ($t$) are separated:

$$\theta(x,y,t) = X(x) Y(y) \Gamma(t)$$

Substituting this ansatz into the heat equation gives:

$$X(x) Y(y) \frac{d\Gamma(t)}{dt} = \alpha \left( Y(y) \Gamma(t) \frac{d^2 X(x)}{dx^2} + X(x) \Gamma(t) \frac{d^2 Y(y)}{dy^2} \right)$$

Dividing both sides by $\alpha X(x) Y(y) \Gamma(t)$:

$$\frac{1}{\alpha \Gamma(t)} \frac{d\Gamma(t)}{dt} = \frac{1}{X(x)} \frac{d^2 X(x)}{dx^2} + \frac{1}{Y(y)} \frac{d^2 Y(y)}{dy^2}$$

Since the left side depends solely on $t$ and the right side depends solely on spatial coordinates $(x,y)$, both sides must equal a negative separation constant, $-\lambda^2$:

$$\frac{1}{\alpha \Gamma(t)} \frac{d\Gamma(t)}{dt} = -\lambda^2$$

$$\frac{1}{X(x)} \frac{d^2 X(x)}{dx^2} + \frac{1}{Y(y)} \frac{d^2 Y(y)}{dy^2} = -\lambda^2$$

Further separating the spatial equation by isolating $x$-dependent terms:

$$\frac{1}{X(x)} \frac{d^2 X(x)}{dx^2} = -\lambda^2 - \frac{1}{Y(y)} \frac{d^2 Y(y)}{dy^2} = -\beta^2$$

This decouples the system into three ordinary differential equations (ODEs):

1. $\dfrac{d^2 X}{dx^2} + \beta^2 X = 0$
2. $\dfrac{d^2 Y}{dy^2} + \gamma^2 Y = 0 \quad \text{where } \gamma^2 = \lambda^2 - \beta^2 \implies \lambda^2 = \beta^2 + \gamma^2$
3. $\dfrac{d\Gamma}{dt} + \alpha \lambda^2 \Gamma = 0$

The Sturm-Liouville problem in the $x$-direction is:

$$X''(x) + \beta^2 X(x) = 0, \quad X(0) = 0, \quad X(L) = 0$$

The general solution is $X(x) = A \cos(\beta x) + B \sin(\beta x)$.

* $X(0) = 0 \implies A = 0$
* $X(L) = 0 \implies B \sin(\beta L) = 0$

For non-trivial solutions ($B \neq 0$), the eigenvalues and corresponding eigenfunctions are:

$$\beta_m = \frac{m \pi}{L}, \quad X_m(x) = \sin\left(\frac{m \pi x}{L}\right) \quad (m = 1, 2, 3, \dots)$$

Similarly, for the $y$-direction:

$$Y''(y) + \gamma^2 Y(y) = 0, \quad Y(0) = 0, \quad Y(H) = 0$$

The eigenvalues and eigenfunctions are:

$$\gamma_n = \frac{n \pi}{H}, \quad Y_n(y) = \sin\left(\frac{n \pi y}{H}\right) \quad (n = 1, 2, 3, \dots)$$

Combining spatial separation constants yields the total eigenvalue $\lambda_{m,n}^2$:

$$\lambda_{m,n}^2 = \beta_m^2 + \gamma_n^2 = \left(\frac{m \pi}{L}\right)^2 + \left(\frac{n \pi}{H}\right)^2$$

Solving the time-dependent ODE for mode $(m,n)$:

$$\frac{d\Gamma_{m,n}(t)}{dt} + \alpha \lambda_{m,n}^2 \Gamma_{m,n}(t) = 0 \implies \Gamma_{m,n}(t) = C_{m,n} \exp\left(-\alpha \lambda_{m,n}^2 t\right)$$

By the principle of superposition, the general transient solution is expressed as a double Fourier series:

$$\theta(x,y,t) = \sum_{m=1}^{\infty} \sum_{n=1}^{\infty} C_{m,n} \sin\left(\frac{m \pi x}{L}\right) \sin\left(\frac{n \pi y}{H}\right) \exp\left[-\alpha \left(\left(\frac{m \pi}{L}\right)^2 + \left(\frac{n \pi}{H}\right)^2\right) t\right]$$

Apply the initial condition at $t = 0$:

$$\theta(x,y,0) = \theta_0 = \sum_{m=1}^{\infty} \sum_{n=1}^{\infty} C_{m,n} \sin\left(\frac{m \pi x}{L}\right) \sin\left(\frac{n \pi y}{H}\right)$$

Using the orthogonality of sine functions over $[0, L]$ and $[0, H]$:

$$\int_0^L \sin\left(\frac{m \pi x}{L}\right) \sin\left(\frac{p \pi x}{L}\right) dx = \frac{L}{2} \delta_{mp}$$

$$\int_0^H \sin\left(\frac{n \pi y}{H}\right) \sin\left(\frac{q \pi y}{H}\right) dy = \frac{H}{2} \delta_{nq}$$

Multiply both sides by $\sin\left(\frac{m \pi x}{L}\right) \sin\left(\frac{n \pi y}{H}\right)$ and integrate across the domain:

$$C_{m,n} = \frac{4}{L H} \int_0^L \int_0^H \theta_0 \sin\left(\frac{m \pi x}{L}\right) \sin\left(\frac{n \pi y}{H}\right) dy \, dx$$

Evaluating the independent 1D integrals:

$$\int_0^L \sin\left(\frac{m \pi x}{L}\right) dx = \frac{L}{m \pi} \left[1 - (-1)^m\right] = \begin{cases} \frac{2L}{m\pi}, & m \text{ odd} \\ 0, & m \text{ even} \end{cases}$$

$$\int_0^H \sin\left(\frac{n \pi y}{H}\right) dy = \begin{cases} \frac{2H}{n\pi}, & n \text{ odd} \\ 0, & n \text{ even} \end{cases}$$

Therefore, the coefficient $C_{m,n}$ is non-zero only for odd values of $m$ and $n$:

$$C_{m,n} = \begin{cases} \frac{16 \theta_0}{\pi^2 m n}, & m, n \text{ both odd} \\ 0, & \text{otherwise} \end{cases}$$

Restoring $T(x,y,t) = T_0 + \theta(x,y,t)$, the exact temperature field is:

$$T(x,y,t) = T_0 + \frac{16 (T_i - T_0)}{\pi^2} \sum_{m=1,3,5\dots}^{\infty} \sum_{n=1,3,5\dots}^{\infty} \frac{1}{m n} \sin\left(\frac{m \pi x}{L}\right) \sin\left(\frac{n \pi y}{H}\right) \exp\left[-\alpha \pi^2 \left( \frac{m^2}{L^2} + \frac{n^2}{H^2} \right) t\right]$$

```python
import numpy as np
import matplotlib.pyplot as plt

def compute_temperature_2d(X, Y, t, L, H, alpha, T0, Ti, num_terms=51):
    """
    Computes the 2D transient temperature field T(x, y, t) using 
    the analytical double Fourier series solution.
    
    Parameters:
    - X, Y        : 2D spatial meshgrid arrays (m)
    - t          : Time instant for evaluation (s)
    - L, H        : Domain width and height (m)
    - alpha      : Thermal diffusivity (m^2/s)
    - T0         : Constant boundary temperature (°C)
    - Ti         : Uniform initial domain temperature (°C)
    - num_terms  : Upper bound truncation for Fourier series sum (odd modes)
    
    Returns:
    - T          : 2D array of temperatures at time t (°C)
    """
    T = np.full_like(X, fill_value=float(T0))
    
    for m in range(1, num_terms, 2):
        for n in range(1, num_terms, 2):
            # Fourier coefficient C_mn
            c_mn = (16.0 * (Ti - T0)) / ((np.pi ** 2) * m * n)
            
            # Spatial mode shapes
            spatial_x = np.sin(m * np.pi * X / L)
            spatial_y = np.sin(n * np.pi * Y / H)
            
            # Exponential decay factor
            decay = np.exp(-alpha * (np.pi ** 2) * ((m / L)**2 + (n / H)**2) * t)
            
            # Accumulate modal contributions
            T += c_mn * spatial_x * spatial_y * decay
            
    return T

L = 1.0          # Domain width in x (m)
H = 1.0          # Domain height in y (m)
alpha = 1e-4     # Thermal diffusivity (m^2/s)
T0 = 20.0        # Homogeneous boundary temperature (°C)
Ti = 100.0       # Initial uniform temperature (°C)
t_eval = 60.0     # Time slice to visualize (s)
Nx, Ny = 150, 150
x = np.linspace(0, L, Nx)
y = np.linspace(0, H, Ny)
X, Y = np.meshgrid(x, y)

T = compute_temperature_2d(X, Y, t=t_eval, L=L, H=H, alpha=alpha, T0=T0, Ti=Ti, num_terms=51)

plt.figure(figsize=(8, 6.5))
contour_filled = plt.contourf(X, Y, T, levels=60, cmap='hot')
colorbar = plt.colorbar(contour_filled)
colorbar.set_label('Temperature $T(x,y,t)$ (°C)', fontsize=12)
contour_lines = plt.contour(X, Y, T, levels=12, colors='black', linewidths=0.5, alpha=0.6)
plt.clabel(contour_lines, inline=True, fontsize=8, fmt='%.1f °C')
plt.title(f'2D Transient Heat Conduction at $t = {t_eval}$ s', fontsize=14, pad=12)
plt.xlabel('Spatial Coordinate $x$ (m)', fontsize=12)
plt.ylabel('Spatial Coordinate $y$ (m)', fontsize=12)
plt.gca().set_aspect('equal', adjustable='box')
plt.grid(True, linestyle='--', alpha=0.3)
plt.tight_layout()

plt.show()
```

```python
import os
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, PillowWriter

def compute_temperature_2d(X, Y, t, L, H, alpha, T0, Ti, num_terms=41):
    """Computes temperature T(x, y, t) using double Fourier series."""
    T = np.full_like(X, fill_value=float(T0))
    for m in range(1, num_terms, 2):
        for n in range(1, num_terms, 2):
            c_mn = (16.0 * (Ti - T0)) / ((np.pi ** 2) * m * n)
            spatial_x = np.sin(m * np.pi * X / L)
            spatial_y = np.sin(n * np.pi * Y / H)
            decay = np.exp(-alpha * (np.pi ** 2) * ((m / L)**2 + (n / H)**2) * t)
            T += c_mn * spatial_x * spatial_y * decay
    return T

L, H = 1.0, 1.0          # Domain dimensions (m)
alpha = 1e-4             # Thermal diffusivity (m^2/s)
T0, Ti = 20.0, 100.0     # Boundary and initial temperatures (°C)
t_start = 0.1            # Start time (s) - small positive float avoids t=0 discontinuity
t_end = (L**2)/alpha/2
num_frames = 60          # Total number of images in sequence
fps = 15                 # Frames per second for video playback
t_series = np.linspace(t_start, t_end, num_frames)

Nx, Ny = 120, 120
x = np.linspace(0, L, Nx)
y = np.linspace(0, H, Ny)
X, Y = np.meshgrid(x, y)

output_dir = "temperature_frames"
os.makedirs(output_dir, exist_ok=True)

levels = np.linspace(T0, Ti, 51)
fig, ax = plt.subplots(figsize=(8, 6.5))
cbar = None

def update(frame_idx):
    global cbar
    ax.clear()
    
    t = t_series[frame_idx]
    T = compute_temperature_2d(X, Y, t, L, H, alpha, T0, Ti)
    
    contour_filled = ax.contourf(X, Y, T, levels=levels, cmap='hot', vmin=T0, vmax=Ti)
    contour_lines = ax.contour(X, Y, T, levels=10, colors='black', linewidths=0.5, alpha=0.5)
    ax.clabel(contour_lines, inline=True, fontsize=8, fmt='%.0f°C')
    
    if cbar is None:
        cbar = fig.colorbar(contour_filled, ax=ax)
        cbar.set_label('$T(x,y,t)$ (°C)', fontsize=12)
        
    ax.set_title(f'$t = {t:.1f}$ s', fontsize=14, pad=12)
    ax.set_xlabel('$x$ (m)', fontsize=12)
    ax.set_ylabel('$y$ (m)', fontsize=12)
    ax.set_aspect('equal', adjustable='box')
    
    frame_path = os.path.join(output_dir, f'frame_{frame_idx:03d}.png')
    plt.savefig(frame_path, dpi=150, bbox_inches='tight')
    
    return contour_filled,

anim = FuncAnimation(fig, update, frames=len(t_series), interval=1000/fps, repeat=False)

gif_filename = "heat_conduction_2d.gif"
anim.save(gif_filename, writer=PillowWriter(fps=fps))

plt.close()

print(f"Sequence saved: {num_frames} PNG frames written to './{output_dir}/'")
print(f"Animation saved: '{gif_filename}'")
```
