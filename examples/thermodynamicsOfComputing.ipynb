# Thermodynamics of Irreversible Computing
## Why is heat generated during computing?
In classical computing, information processing is an irreversible process. According to **Landauer's Principle**, erasing one bit of information releases at least $k_B T \ln 2$ of heat into the environment.

### Quantifying the limit (J/FLOP)

```python
import matplotlib.pyplot as plt
import numpy as np

# Constants
kB = 1.38e-23  # J/K
T = 300        # K
landauer_limit_joules_per_bit = kB * T * np.log(2)

print(f'Landauer Limit (J/bit): {landauer_limit_joules_per_bit:.2e}')
```

### Comparison with Modern Systems

```python
modern_values = np.logspace(-15, -10, 50)

plt.figure(figsize=(8, 6))
plt.semilogy(modern_values, [landauer_limit_joules_per_bit]*len(modern_values), 'r--', label='Landauer Limit (Theoretical)')
plt.semilogy(modern_values, modern_values, 'b-', label='Modern Computing (Approximate)')

plt.xlabel('Energy per bit flip [J]')
plt.ylabel('Energy [J]')
plt.title('Thermodynamic Limit vs Modern Computing')
plt.grid(True, which="both", ls="-", alpha=0.5)
plt.legend()
plt.show()
```
