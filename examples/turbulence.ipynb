**1. Turbulence Model Selection: Realizable $k-\epsilon$ vs. $k-\omega$ SST**

Selecting an appropriate Reynolds-Averaged Navier-Stokes (RANS) closure model balances computational efficiency against boundary layer resolution in room-scale airflow simulations.

| Specification | Realizable / RNG $k-\epsilon$ | $k-\omega$ SST (Shear Stress Transport) |
| --- | --- | --- |
| **Wall Grid Treatment** | Standard wall functions ($30 < y^+ < 300$) | Viscous sublayer resolved ($y^+ \approx 1$) |
| **Jet & Plume Spreading** | High accuracy for buoyant plumes & tile jets | Tendency to over-predict eddy viscosity in far-field |
| **Flow Separation** | Moderate accuracy under adverse pressure gradients | High accuracy near rack edges & complex geometries |
| **Mesh Requirements** | Low-to-moderate grid count per unit volume | Extremely high grid density (boundary inflation layers) |
| **Primary DC Application** | **Macro-scale room CFD** (data halls, aisles) | **Micro-scale rack CFD** (heatsinks, cold-plates) |

* **Realizable $k-\epsilon$ Recommendation:** Best suited for room-scale thermal airflow. It captures buoyant thermal plumes and tile discharge jets without requiring millions of near-wall boundary cells.
* **$k-\omega$ SST Recommendation:** Recommended for detailed rack-chassis airflow studies, severe flow-separation areas, or fan blade thermal analysis where viscous boundary layers must be explicitly resolved.

**2. Room-Scale Boundary Condition Definitions**

Accurate room-scale models require specifying conservation variables at domain boundaries to match physical server room operation.

* **Velocity Inlets (CRAH Discharge / Perforated Tile Surfaces):**
* **Specified Parameters:** Velocity magnitude/vector $\mathbf{u} = (0, 0, w_{\text{in}})$, cold supply temperature $T_{\text{supply}}$, turbulence intensity $I = 5\text{--}10\%$, and turbulent viscosity ratio $\mu_t / \mu \approx 10$.
* **Application:** Used at cooling unit supply openings or individual floor tile faces when airflow distribution is fixed.


* **Pressure Outlets (CRAH Return / Plenum Intakes):**
* **Specified Parameters:** Fixed static gauge pressure $p_{\text{gauge}} = 0\text{ Pa}$, specified backflow total temperature $T_{\text{backflow}}$ to preserve thermal numerical stability if local reverse flow occurs.
* **Application:** Applied at return air grilles or ceiling plenum suction boundaries.


* **Volumetric Heat Sources (Server Rack Blocks):**
* **Specified Parameters:** Volumetric thermal source term $q''' = \frac{P_{\text{rack}}}{V_{\text{rack}}} \quad [\text{W/m}^3]$, where $P_{\text{rack}}$ is active IT power consumption.
* **Application:** Homogenized across the porous rack volume alongside fan-curve momentum source terms.


* **Wall Boundaries (Containment Structures, Floor, Ceiling):**
* **Hydrodynamic:** No-slip condition ($\mathbf{u} = \mathbf{0}$) with standard wall functions.
* **Thermal:** Internal containment panels are modeled as adiabatic ($\nabla T \cdot \mathbf{n} = 0$). Outer room envelope walls incorporate ambient convective flux:

$$-k \frac{\partial T}{\partial n} = h_{\text{ext}}(T_{\text{wall}} - T_{\text{ambient}})$$
