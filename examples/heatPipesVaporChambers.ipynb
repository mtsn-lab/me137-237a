# Heat Pipes and Vapor Chambers

Heat pipes and vapor chambers are passive two-phase thermal management devices that utilize phase change (evaporation and condensation) to transport and distribute heat at high effective thermal conductivities. A heat pipe uses capillary action to transport heat axially along a linear path from a heat source to a displaced heat sink. A vapor chamber spreads heat planar-wise from a concentrated high-flux source (e.g., CPU/GPU die) to a larger surface area to minimize localized thermal resistance and hot spots. Heat absorbed at the evaporator vaporizes the internal working fluid; vapor flows driven by pressure gradients to the condenser, releases latent heat, and returns as liquid to the evaporator through a porous wick structure via capillary pumping.

A standard heat pipe consists of three operational regions along its length $L$: an evaporator section ($L_e$), an adiabatic section ($L_a$), and a condenser section ($L_c$). To model the fluid transport analytically, we define an effective transport length $L_{\text{eff}}$:


$$L_{\text{eff}} = \frac{1}{2} L_e + L_a + \frac{1}{2} L_c$$

For continuous steady-state operation, the maximum capillary pressure generated by the wick ($\Delta p_{c,\max}$) must equal or exceed the total pressure drop experienced by the circulating fluid across the liquid and vapor paths:

$$\Delta p_{c,\max} \ge \Delta p_l + \Delta p_v + \Delta p_g$$

where $\Delta p_l$ is the pressure drop in the liquid phase moving through the wick, $\Delta p_v$ is the pressure drop in the vapor phase flowing through the core, and $\Delta p_g$ is the hydrostatic head due to gravity ($\Delta p_g = \rho_l g L \sin\psi$, where $\psi$ is the inclination angle relative to horizontal). We can formulate the capillary pressure using the Young-Laplace equation across a spherical liquid-vapor meniscus formed in a wick pore of effective radius $r_e$ with contact angle $\theta$:

$$\Delta p_{c,\max} = \frac{2\sigma \cos\theta}{r_e}$$

where $\sigma$ is the surface tension of the working fluid. The liquid flow through the porous wick matrix is governed by Darcy's Law. For a local mass flow rate $\dot{m}(x)$:

$$\frac{dp_l}{dx} = -\frac{\mu_l}{\rho_l K A_w} \dot{m}(x)$$

where $\mu_l$ is liquid dynamic viscosity, $\rho_l$ is liquid density, $K$ is wick permeability, and $A_w$ is the cross-sectional area of the wick. Relating mass flow rate to total axial heat transport $Q$ via the latent heat of vaporization $h_{fg}$ ($\dot{m} = Q / h_{fg}$), integration over $L_{\text{eff}}$ yields:

$$\Delta p_l = \frac{\mu_l Q L_{\text{eff}}}{K A_w \rho_l h_{fg}}$$

Assuming low-velocity, laminar, incompressible vapor flow inside a circular vapor core of radius $r_v$, Hagen-Poiseuille flow gives:

$$\frac{dp_v}{dx} = -\frac{8 \mu_v \dot{m}(x)}{\pi \rho_v r_v^4}$$

Integrating along $L_{\text{eff}}$ with $\dot{m} = Q / h_{fg}$:

$$\Delta p_v = \frac{8 \mu_v Q L_{\text{eff}}}{\pi \rho_v r_v^4 h_{fg}}$$

Substituting pressure drop expressions into the capillary balance condition ($\Delta p_{c,\max} = \Delta p_l + \Delta p_v$, assuming horizontal operation $\Delta p_g = 0$):

$$\frac{2\sigma \cos\theta}{r_e} = \left( \frac{\mu_l}{K A_w \rho_l h_{fg}} + \frac{8 \mu_v}{\pi \rho_v r_v^4 h_{fg}} \right) Q_{\max,c} L_{\text{eff}}$$

Solving for maximum capillary heat rate $Q_{\max,c}$:

$$Q_{\max,c} = \frac{\frac{2 \sigma \cos\theta}{r_e}}{L_{\text{eff}} \left( \frac{\mu_l}{K A_w \rho_l h_{fg}} + \frac{8 \mu_v}{\pi \rho_v r_v^4 h_{fg}} \right)}$$

Because the liquid viscous resistance dominates over the vapor resistance ($\Delta p_l \gg \Delta p_v$) in typical commercial heat pipes, neglecting vapor pressure loss simplifies the limit to:

$$Q_{\max,c} \approx \left( \frac{\rho_l \sigma h_{fg}}{\mu_l} \right) \cdot \left( \frac{2 K A_w \cos\theta}{r_e L_{\text{eff}}} \right)$$

The term $\frac{\rho_l \sigma h_{fg}}{\mu_l}$ is defined as the fluid figure of merit, allowing direct liquid selection ranking based purely on thermophysical property combinations. The total thermal resistance of a heat pipe ($R_{\text{th}} = \frac{T_e - T_c}{Q}$) is modeled as a series network of individual conductive and phase-change resistances:

$$R_{\text{th}} = R_{\text{wall},e} + R_{\text{wick},e} + R_{\text{inter},e} + R_v + R_{\text{inter},c} + R_{\text{wick},c} + R_{\text{wall},c}$$

where the wall conduction resistance is:
$$R_{\text{wall}} = \frac{\ln(r_o / r_i)}{2 \pi k_{\text{wall}} L}$$

and the effective wick reistance is:

$$R_{\text{wick}} = \frac{\ln(r_i / r_v)}{2 \pi k_{\text{eff}} L}$$

For a liquid-saturated sintered metal powder wick, the effective thermal conductivity $k_{\text{eff}}$ can be estimated as:

$$k_{\text{eff}} = k_l \left[ \frac{2 k_l + k_s - 2\phi(k_l - k_s)}{2 k_l + k_s + \phi(k_l - k_s)} \right]$$

where $\phi$ is wick porosity, $k_s$ is solid wick conductivity, and $k_l$ is liquid conductivity.

The overall heat transport capacity of a heat pipe is bounded by five physical limitations:

1. **Capillary Limit:** Maximum pressure pumping capability of the wick structure is exceeded, leading to evaporator dryout.
2. **Boiling Limit:** Excessive radial heat flux triggers nucleate boiling within the evaporator wick, forming insulating vapor bubbles.
3. **Entrainment Limit:** High vapor shear stress strips liquid droplets from the porous wick surface and carries them back to the condenser.
4. **Sonic Limit:** Vapor flow velocity reaches Mach 1 at the evaporator exit, choking mass flow.
5. **Viscous Limit:** Low operating temperatures create high vapor viscosity and low vapor pressure, restricting vapor flow.
