**1. Airflow Management Topologies & Containment Modeling**

Data hall airflow patterns are dictated by physical boundary configurations designed to prevent the mixing of cold supply air and hot exhaust air.

* **Uncontained / Open Aisle Configuration:**
* **CFD Representation:** Open boundaries between aisles allow unconstrained momentum and thermal mixing across the room height.
* **Physical Effect:** High propensity for hot air recirculation over rack tops and cold supply bypass directly back to CRAH units.


* **Cold Aisle Containment (CAC):**
* **CFD Representation:** Physical doors and ceiling panels above the cold aisle are modeled as zero-thickness wall boundaries with no-slip and adiabatic conditions.
* **Pressure Dynamics:** Cold aisle operates at positive static pressure relative to the room ($\Delta p > 0$), forcing cold air through server chassis momentum sinks.


* **Hot Aisle Containment (HAC):**
* **CFD Representation:** Encloses hot aisle with physical wall boundaries and routes hot exhaust directly into a ceiling return plenum using pressure-driven boundary conditions.
* **Thermal Impact:** Allows higher CRAH return air temperatures ($\Delta T_{\text{return}} \uparrow$), maximizing cooling coil heat exchange efficiency.

**2. Fluid Dynamic & Thermal Short-Circuiting Phenomena**

Imperfections in data center airflow management lead to two primary operational inefficiencies:

```
[ CRAH Supply ] ──( Cold Air )──> [ Bypass Airflow ] ───────────────┐
                                                                    v
[ Server Exhaust ] ──( Hot Air )──> [ Thermal Recirculation ] ──> [ CRAH Return ]
                                            │
                                            └─> [ Server Inlet ] (Hot Spot!)

```

* **Bypass Airflow (Cold Air Short-Circuiting):**
* **Definition:** Condition where conditioned supply air returns directly to the CRAH unit without absorbing thermal energy from active IT equipment.
* **CFD Identification:** High-velocity, low-temperature streamlines originating at perforated tiles or open floor cutouts that enter return grilles without passing through server porous zones.
* **Cause:** Over-provisioning airflow rate relative to total IT fan demand ($\dot{V}_{\text{CRAH}} > \sum \dot{V}_{\text{rack}}$) or unsealed cable grommets.


* **Thermal Recirculation (Hot Air Short-Circuiting):**
* **Definition:** Condition where hot rack exhaust air recirculates back into the inlet of the same or neighboring server racks.
* **CFD Identification:** Reversed streamline trajectories near rack tops/sides and elevated local temperature fields ($T_{\text{inlet}} \gg T_{\text{supply}}$) at upper server rack units.
* **Cause:** Under-provisioning supply air, unsealed empty rack spaces (lack of blanking panels), or uncontained aisle architectures.

**3. Volumetric Governing Terms for Recirculation & Bypass**

In CFD post-processing, these phenomena are quantified by integrating local velocity and temperature vectors across key subdomains:

* **Inlet Recirculation Energy Flux ($Q_{\text{recirc}}$):**
$$Q_{\text{recirc}} = \int_{A_{\text{inlet}}} \rho c_p \max(T(\mathbf{r}) - T_{\text{supply}}, \, 0) \, [\mathbf{u}(\mathbf{r}) \cdot \mathbf{n}] \, dA$$


* **Bypass Mass Flow Rate ($\dot{m}_{\text{bypass}}$):**
$$\dot{m}_{\text{bypass}} = \int_{A_{\text{return}}} \rho \, \delta_{\text{bypass}}(\mathbf{r}) \, [\mathbf{u}(\mathbf{r}) \cdot \mathbf{n}] \, dA$$


where $\delta_{\text{bypass}}(\mathbf{r}) = 1$ if $T(\mathbf{r}) \le T_{\text{supply}} + \epsilon$, and $0$ otherwise.
