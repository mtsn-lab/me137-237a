#  Formulate the incompressible Navier-Stokes and energy equations for thermal airflow in room-scale enclosures

For room-scale thermal airflow in data centers, air is modeled as an incompressible fluid using the Boussinesq approximation to capture buoyancy-driven convective flows generated by server thermal loads. We first apply the conservation of mass principle. Assuming a constant reference density $\rho_0$, we can write:

$$\nabla \cdot \mathbf{u} = 0$$


We then apply the principle of conservation of momentum, which in our case is the incompressible Navier-Stokes equation with the Boussinesq Approximation. This approximation is valid when density variations are small ($\beta \Delta T \ll 1$), which holds well for typical room-scale temperature differentials ($\Delta T \approx 10\text{–}20\text{ °C}$). Thermal buoyancy is represented as a body force proportional to local temperature variations relative to a reference temperature $T_0$. We can then write:

$$\frac{\partial \mathbf{u}}{\partial t} + (\mathbf{u} \cdot \nabla)\mathbf{u} = -\frac{1}{\rho_0}\nabla p^* + \nu \nabla^2 \mathbf{u} - \mathbf{g}\beta(T - T_0)$$

where $p^* = p - \rho_0 \mathbf{g}\cdot\mathbf{r}$ is the dynamic pressure excluding the hydrostatic head.

Finally, we can apply the conservation of energy principle. We can express this in terms of temperature $T$, assuming constant specific heat $c_p$ and thermal conductivity $k$, while incorporating a volumetric heat source $q'''$ to model dissipated server power:

$$\frac{\partial T}{\partial t} + \mathbf{u} \cdot \nabla T = \alpha \nabla^2 T + \frac{q'''}{\rho_0 c_p}$$

where $\alpha = \frac{k}{\rho_0 c_p}$ is the thermal diffusivity of air.

**Nomenclature & Variable Definitions**

| Symbol | Variable Description | Standard Units |
| --- | --- | --- |
| $\mathbf{u}$ | Velocity vector field $(u, v, w)$ | m/s |
| $p^*$ | Modified dynamic pressure | Pa |
| $T$ | Static air temperature field | K or °C |
| $T_0$ | Reference ambient temperature | K or °C |
| $\rho_0$ | Reference air density evaluated at $T_0$ | $\text{kg/m}^3$ |
| $\nu$ | Kinematic viscosity ($\mu/\rho_0$) | $\text{m}^2/\text{s}$ |
| $\alpha$ | Thermal diffusivity ($k/\rho_0 c_p$) | $\text{m}^2/\text{s}$ |
| $\beta$ | Thermal expansion coefficient ($\approx 1/T_0$ for ideal gas) | 1/K |
| $\mathbf{g}$ | Gravitational acceleration vector $(0, 0, -g)$ | $\text{m/s}^2$ |
| $q'''$ | Volumetric heat generation rate | $\text{W/m}^3$ |
