# Cooling distribution units

* **Per-GPU Heat Dissipation:** A single NVIDIA Rubin GPU dissipates up to **~2,400 W** (~2.4 kW).
* **Rack-Level Thermal Density:**
    * **NVIDIA Vera Rubin NVL72 Rack:** Houses 72 Rubin GPUs and 36 Vera CPUs across 18 compute trays (plus 9 NVLink 6 switch trays) in a single rack domain.
    * **Heat Generation / Power Consumption:** **>200 kW per rack**, with next-generation roadmap configurations expanding toward **500–600 kW per rack**. (By comparison, the prior-generation Grace Blackwell GB200 NVL72 draws ~120–140 kW per rack).


* **Direct Liquid Heat Capture Ratio (HCR):** Over **92% to 95%** of the rack’s total thermal output is removed directly via liquid cold plates attached to GPUs, CPUs, and NVLink switch silicon, leaving minimal residual heat for air cooling.


CDU architectures for Vera Rubin deployments depend on the data center layout:


* **Warm Water Direct Liquid Cooling (DLC):**
* **Supply Temperature ($T_{supply}$):** $\approx 45^\circ\text{C}$ ($113^\circ\text{F}$).
* **Return Temperature ($T_{return}$):** $\approx 55^\circ\text{C}\text{ to }65^\circ\text{C}$ ($131^\circ\text{F}\text{ to }149^\circ\text{F}$).


* **Flow Rates:**
    * **Per-Rack Flow Rate:** **30 to 50 L/min** per NVL72 rack under nominal load, ramping up to **~130 L/min** under peak thermal demand.
    * **Multi-Rack In-Row CDU (1.8 MW Unit):** Handles aggregate flow rates of **300 to 1,000+ L/min** (~80 to 260+ GPM) across the primary-to-secondary plate heat exchangers.


* **Operating Pressure:** Designed to operate up to **5 bar** in accordance with OCP (Open Compute Project) / MGX manifold specifications.
* **Cooling Medium:** The Technology Cooling System (TCS) secondary loop utilizes treated deionized water or a 25% propylene glycol mix (PG25) with anti-corrosion inhibitors.
