Macro-scale CFD models room-scale environments by substituting complex physical micro-geometries (such as perforated tiles, server heat sinks, and cooling coils) with continuous porous zones that exert equivalent momentum sink forces on the fluid.

**1. Mathematical Scaling: From Darcy's Law to Darcy-Forchheimer**

* **Darcy's Law (Laminar Creeping Flow):**
For low-velocity, viscous-dominated flows ($\text{Re} < 1$), momentum loss per unit length is linearly proportional to fluid velocity:
$$\nabla p = -\frac{\mu}{\alpha} \mathbf{u}$$


* **Darcy-Forchheimer Formulation (High Velocity / Turbulent Flow):**
At data center airflow speeds ($1\text{--}5\text{ m/s}$), inertial form-drag effects caused by micro-scale flow separation become significant. Adding a quadratic inertial term yields the Darcy-Forchheimer momentum sink source term $S_i$ added to the Navier-Stokes momentum equation:
$$S_i = -\left( \frac{\mu}{\alpha_i} u_i + \frac{1}{2} C_{2,i} \rho \vert{}\mathbf{u}\vert{} u_i \right)$$


In full 3D tensor notation for anisotropic media:
$$\mathbf{S} = -\left( \mathbf{D} \mu \mathbf{u} + \mathbf{C} \frac{1}{2} \rho \vert{}\mathbf{u}\vert{} \mathbf{u} \right)$$



---

**2. Deriving Coefficients from Experimental Pressure-Drop Curves**

Manufacturer data for tiles, coils, and server chassis provides pressure drop ($\Delta p$) versus face velocity ($u$). This relationship is fitted to a quadratic polynomial over the component thickness $\Delta x$:

$$\Delta p = a u + b u^2 \quad \implies \quad \frac{\Delta p}{\Delta x} = \left(\frac{a}{\Delta x}\right) u + \left(\frac{b}{\Delta x}\right) u^2$$

Equating polynomial coefficients to the Darcy-Forchheimer formulation yields:

$$\alpha = \frac{\mu \Delta x}{a}, \qquad C_2 = \frac{2 b}{\rho \Delta x}$$

**3. Modeling Enclosure Micro-Geometries**

**Perforated Floor Tiles**

* Modeled as thin porous surfaces or 1D porous jump boundaries.
* **Inertial Resistance ($C_2$):** Dominates due to sharp-edged orifice pressure losses. Calculated using tile open-area ratio $\sigma$:

$$K_L = \left(\frac{1 - \sigma}{\sigma^2}\right) \quad \implies \quad C_2 = \frac{K_L}{\Delta x}$$



**Server Racks**

* Modeled as volumetric 3D anisotropic blocks.
* **Streamwise Axis ($x$):** Calibrated to match internal server fan curve airflow vs. pressure loss characteristics.
* **Transverse Axes ($y, z$):** Assigned artificial high-resistance coefficients ($C_{2,y}, C_{2,z} \gg C_{2,x}$) to force 1D unidirectional flow through the server chassis and prevent non-physical internal cross-flow.
* **Thermal Source:** Coupled with a volumetric heat generation term $q''' = \frac{P_{\text{rack}}}{V_{\text{rack}}}$.

**Cooling Coils (CRAH Units)**

* Modeled as isotropic or planar-anisotropic porous volumes.
* High viscous resistance ($\alpha^{-1}$) models micro-channel skin friction; high inertial resistance ($C_2$) models fin-tube flow tortuosity.
* Coupled with a negative energy sink term ($q''' < 0$) or effectiveness model to simulate air temperature drop.

**4. Parameter Nomenclature**

| Symbol | Parameter Description | Standard Units |
| --- | --- | --- |
| $S_i$ | Volumetric momentum sink vector in direction $i$ | $\text{N/m}^3$ |
| $\alpha$ | Permeability of the porous medium | $\text{m}^2$ |
| $1/\alpha$ | Viscous resistance coefficient | $\text{1/m}^2$ |
| $C_2$ | Inertial resistance coefficient | $\text{1/m}$ |
| $\mathbf{D}, \mathbf{C}$ | Diagonal matrices for directional viscous and inertial resistances | $\text{1/m}^2, \text{1/m}$ |
| $\Delta x$ | Physical thickness of the porous zone along flow direction | m |
| $K_L$ | Non-dimensional pressure loss coefficient | — |
| $\sigma$ | Perforated tile open-area ratio ($A_{\text{holes}} / A_{\text{total}}$) | — |
