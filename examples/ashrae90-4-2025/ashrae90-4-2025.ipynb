# ASHRAE Standard 90.4-2025

## Annualized Mechanical Load Component
![equation6-5](equation-6-5.png)

![table-6-5](table-6-5.png)

In ASHRAE Standard 90.4, **Annualized MLC** and **Design ELC** are dimensionless efficiency metrics that evaluate energy losses relative to the IT equipment load.

**Annualized Mechanical Load Component ($\text{MLC}_{\text{annualized}}$)**

The Annualized MLC represents the ratio of total annual energy consumed by all mechanical cooling infrastructure to the total annual energy delivered to the IT equipment.

$$\text{MLC}_{\text{annualized}} = \frac{E_{\text{mech, annual}}}{E_{\text{IT, annual}}} = \frac{\sum_{k=1}^{N} P_{\text{mech}, k} \cdot \Delta t_k}{\sum_{k=1}^{N} P_{\text{IT}, k} \cdot \Delta t_k}$$

Where:

* $E_{\text{mech, annual}}$ = Total annual mechanical energy consumption ($\text{kWh}$).
* $E_{\text{IT, annual}}$ = Total annual IT equipment energy consumption ($\text{kWh}$).
* $P_{\text{mech}, k}$ = Combined power consumption of all mechanical infrastructure (chillers, pumps, CRAC/CRAH fans, heat rejection equipment, and humidification) during time or climate bin $k$ ($\text{kW}$).
* $P_{\text{IT}, k}$ = IT equipment load during time or climate bin $k$ ($\text{kW}$).
* $\Delta t_k$ = Duration of time or climate bin $k$ ($\text{hours}$).
* $N$ = Total number of time intervals in a full calendar year ($8,760\text{ hours}$) or climate/load bins.

## Design Electrical Loss Component (Design ELC)

![table-8-5](table-8-5.png)

**Design Electrical Loss Component ($\text{ELC}_{\text{design}}$)**

The Design ELC quantifies electrical distribution losses as a fraction of the IT load served. It is evaluated at specified operating load points ($100\%$, $75\%$, $50\%$, and $25\%$ of design IT capacity).

$$\text{ELC} = \frac{P_{\text{loss, total}}}{P_{\text{IT}}} = \frac{\sum_{j=1}^{m} P_{\text{loss}, j}}{P_{\text{IT}}}$$

Where:

* $P_{\text{loss, total}}$ = Total power lost across all electrical distribution equipment upstream of the IT loads ($\text{kW}$).
* $P_{\text{loss}, j}$ = Power loss in component segment $j$ ($\text{kW}$).
* $P_{\text{IT}}$ = Design IT equipment load at the evaluated operating point ($\text{kW}$).
* $m$ = Total number of electrical distribution segments in series.

## Calculating Annualized MLC from a Thermofluid Flow Network Model.

Let's assume that we have established the thermofluid network equations. We model the airside, waterside, and refrigerant loops as a network of nodes (pressures $p$, temperatures $T$) connected by flow paths with hydraulic/pneumatic resistances ($R$). The IT rack thermal load (heat source) is then:

$$Q_{\text{IT}} = \dot{m}_{\text{air}} \cdot c_{p, \text{air}} \cdot (T_{\text{return}} - T_{\text{supply}})$$

The air-side and water-side flow resistances are:
$$\Delta p = R \cdot \dot{V}^2$$

The heat exchangers, such as the CRAH/CRACs and cooling towers can be modeled as:
$$Q = \epsilon \cdot C_{\text{min}} \cdot (T_{\text{fluid, in}} - T_{\text{fluid, out}})$$

We can now define the mechanical power draw. For every fluid-moving element in the network, we define the instantaneous electrical power consumption ($P$) as a function of volumetric flow rate ($\dot{V}$), pressure drop ($\Delta P$), and overall efficiency ($\eta$). The fan and pump power draw is:

$$P_{\text{fan/pump}} = \frac{\dot{V} \cdot \Delta P}{\eta_{\text{fluid}} \cdot \eta_{\text{motor}} \cdot \eta_{\text{VFD}}}$$

The chiller power draw is then:
$$P_{\text{chiller}} = \frac{Q_{\text{cooling}}}{\text{COP}(T_{\text{condenser, in}}, \text{PLR})}$$

where $\text{PLR}$ is the part load ratio and $\text{COP}$ is calculated dynamically based on condenser entering water/air temperature. We then run the network simulations over different climate intervals. We impose outdoor environmental boundary conditions (dry-bulb $T_{\text{db}}$, wet-bulb $T_{\text{wb}}$) from typical meteorological year weather data for each hour $t = 1 \dots 8760$.

At each hour, we solve the conservation of mass, momentum, and energy across the network to determine the operational state of the economizer (free cooling, partial free cooling, or mechanical DX/chiller cooling) and evaluate total instantaneous mechanical power:

$$P_{\text{mech}}(t) = \sum P_{\text{fans}}(t) + \sum P_{\text{pumps}}(t) + \sum P_{\text{chillers}}(t) + \sum P_{\text{heat\_rejection}}(t) + \sum P_{\text{humidification}}(t)$$

We then integrate the total mechanical power and IT load across all 8,760 hours to compute the annualized MLC:

$$\text{MLC}_{\text{annualized}} = \frac{\sum_{t=1}^{8760} P_{\text{mech}}(t) \cdot \Delta t}{\sum_{t=1}^{8760} P_{\text{IT}}(t) \cdot \Delta t}$$

Using a 1D thermo-fluid network model (built in environments such as Modelica, MATLAB/Simulink, Python) is the standard engineering methodology for determining annualized MLC.
